```python
# ============================
# 1. Importera bibliotek
# ============================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Snyggare grafer
sns.set(style="whitegrid")

# ============================
# 2. Ladda datasetet
# ============================

df = pd.read_csv("../data/telco.csv")

# Visa de första raderna
df.head()
```

```python
# ============================
# Datasetets struktur och datatyper
# ============================

df.info()
```

```python
# ============================
# Kontrollera missing values
# ============================

df.isnull().sum()
```

```python
# ============================
# Statistisk sammanfattning för numeriska kolumner
# ============================

df.describe()
```

```python
# ============================
# Statistisk sammanfattning för kategoriska kolumner
# ============================

df.describe(include="object")
```

```python
# ============================
# Fördelning av Churn (Yes/No)
# ============================

plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Churn", palette="viridis")
plt.title("Churn distribution")
plt.show()
```

```python
# ============================
# Fördelning av gender
# ============================

plt.figure(figsize=(6,4))
sns.countplot(data=df, x="gender", palette="magma")
plt.title("Gender distribution")
plt.show()
```

```python
# ============================
# Churn per gender
# ============================

plt.figure(figsize=(6,4))
sns.countplot(data=df, x="gender", hue="Churn", palette="coolwarm")
plt.title("Churn by gender")
plt.show()
```

```python
# ============================
# Churn per InternetService
# ============================

plt.figure(figsize=(7,4))
sns.countplot(data=df, x="InternetService", hue="Churn", palette="coolwarm")
plt.title("Churn by Internet Service Type")
plt.show()
```

```python
# ============================
# Distribution av tenure (hur länge kunden varit kund)
# ============================

plt.figure(figsize=(8,4))
sns.histplot(df["tenure"], kde=True, bins=30, color="teal")
plt.title("Distribution of Tenure")
plt.show()
```

```python
# ============================
# Boxplot för att se skillnader i MonthlyCharges mellan churn/no churn
# ============================

plt.figure(figsize=(8,4))
sns.boxplot(data=df, x="Churn", y="MonthlyCharges", palette="coolwarm")
plt.title("Monthly Charges by Churn")
plt.show()
```

```python
# ============================
# Korrelationer mellan numeriska variabler
# ============================

# Välj endast numeriska kolumner
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(10,6))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap (Numeric Features Only)")
plt.show()

```

```python
# ============================
# Pairplot för numeriska features (kan ta lite tid)
# ============================

sns.pairplot(df[["tenure", "MonthlyCharges", "TotalCharges", "Churn"]], hue="Churn")
plt.show()
```
